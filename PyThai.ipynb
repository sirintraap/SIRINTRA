```python
from pythainlp.tokenize import word_tokenize
text=''
c=word_tokenize(text,engine='mm')
c
```

```python
import numpy as np
import pandas as pd

words = c

pd.value_counts(np.array(words))
df = pd.value_counts(np.array(words))

print('Words:', df.index)
print('Values:', df.values)
```
